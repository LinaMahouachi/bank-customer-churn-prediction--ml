```python
df.isna().sum().to_frame()
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("messy_data.csv")

print("========== INITIAL DATA ==========")
print(df.info())
print(df.head())


# =========================
# 1. DUPLICATES
# =========================
print("\n========== DUPLICATES ==========")
duplicates = df.duplicated().sum()
print("Number of duplicate rows:", duplicates)

# FIX
df = df.drop_duplicates()


# =========================
# 2. WRONG DATA TYPES
# =========================
print("\n========== DATA TYPES BEFORE ==========")
print(df.dtypes)

# Detect problematic numeric columns
print("\nSample Age values:", df['Age'].head())
print("Sample Balance values:", df['Balance'].head())

# FIX
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Balance'] = pd.to_numeric(df['Balance'], errors='coerce')

print("\n========== DATA TYPES AFTER ==========")
print(df.dtypes)


# =========================
# 3. MISSING VALUES
# =========================
print("\n========== MISSING VALUES BEFORE ==========")
print(df.isnull().sum())

# FIX
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

df[num_cols] = df[num_cols].fillna(df[num_cols].median())

for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("\n========== MISSING VALUES AFTER ==========")
print(df.isnull().sum())


# =========================
# 4. CATEGORICAL ISSUES (Gender)
# =========================
print("\n========== GENDER VALUES BEFORE ==========")
print(df['Gender'].unique())

# FIX
df['Gender'] = df['Gender'].replace({
    'F': 'Female',
    'M': 'Male'
})

print("\n========== GENDER VALUES AFTER ==========")
print(df['Gender'].unique())


# =========================
# 5. OUTLIERS
# =========================
print("\n========== OUTLIERS CHECK (AGE) ==========")
print("Min Age:", df['Age'].min())
print("Max Age:", df['Age'].max())

print("\n========== OUTLIERS CHECK (BALANCE) ==========")
print("Min Balance:", df['Balance'].min())
print("Max Balance:", df['Balance'].max())

# FIX using IQR
def remove_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return df[(df[col] >= lower) & (df[col] <= upper)]

df = remove_outliers(df, 'Age')
df = remove_outliers(df, 'Balance')

print("\nAfter removing outliers:")
print("Max Age:", df['Age'].max())
print("Max Balance:", df['Balance'].max())


# =========================
# FINAL CHECK
# =========================
print("\n========== FINAL DATA CHECK ==========")
print(df.info())
print("\nRemaining missing values:\n", df.isnull().sum())
df.head()
```
